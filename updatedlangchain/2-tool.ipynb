```python
from langchain.chat_models import init_chat_model

model = init_chat_model(
    "openai/gpt-oss-20b",
    model_provider="groq"
)

response = model.invoke("Explain LangGraph in simple words.")

print(response.content)
```

```python
from langchain.tools import tool
@tool
def get_weather(location:str)->str:
    """Get the weather at a location"""
    return f"It's sunny in {location}"

model_with_tool=model.bind_tools([get_weather])
```

```python
response=model_with_tool.invoke("what is weather in bhubaneswar")
```

```python
response
```

```python
tools = [get_weather]

tool_map = {tool.name: tool for tool in tools}

response = model_with_tool.invoke("What is the weather in New York?")

for res in response.tool_calls:
    tool_name = res["name"]
    arguments = res["args"]

    tool = tool_map[tool_name]
    result = tool.invoke(arguments)

    print(result)
```
